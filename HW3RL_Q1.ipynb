```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
tol = 1e-8    #tolerance of when to stop iterating
max_iter = 1000      #number of iterations
```

```python
def f_a(x):
  return x**2 + 3*x - 1

x0_list_a = np.linspace(-10, 10, 10)
```

```python
def df_a(x):
  return 2*x + 3
```

```python
def f_b(x):
  return x*np.sin(2*x)


x0_list_b = np.linspace(1, 10, 20)
```

```python
def df_b(x):
  return np.sin(2*x)+2*x*np.cos(2*x)
```

```python
def f_c(x):
  return 2*x**3 + 3*x**2 +2*np.sin(np.cos(x)) - 3


x0_list_c = np.linspace(-1, 1, 100)
```

```python
def df_c(x):
  return 6*x**2 + 6*x - 2*np.cos(np.cos(x))*np.sin(x)
```

```python
def newton_alg(f, df, x0, tol, max_iter):
  x = x0
  history = [x]
  for k in range(max_iter):
    x = x - f(x)/df(x)
    history.append(x)
    if abs(f(x)) < tol:
      return x, k + 1, True, history    #finished
  return x, max_iter, False, history     #didn't pass
```

```python
for x0 in x0_list_a:
  x_final, n_iter, converged, hist = (newton_alg(f_a, df_a, x0, tol, max_iter))
  print(
    f"x0 = {x0:6.2f} | root = {x_final:.8f} | "
    f"f(root) = {f_a(x_final):.2e} | "
    f"iters = {n_iter} | converged = {converged}"
)
```

```python
for x0 in x0_list_b:
  x_final, n_iter, converged, hist = (newton_alg(f_b, df_b, x0, tol, max_iter))
  print(
    f"x0 = {x0:6.2f} | root = {x_final:.8f} | "
    f"f(root) = {f_b(x_final):.2e} | "
    f"iters = {n_iter} | converged = {converged}"
)
```

```python
for x0 in x0_list_c:
  x_final, n_iter, converged, hist = (newton_alg(f_c, df_c, x0, tol, max_iter))
  print(
    f"x0 = {x0:6.2f} | root = {x_final:.8f} | "
    f"f(root) = {f_c(x_final):.2e} | "
    f"iters = {n_iter} | converged = {converged}"
)
```
